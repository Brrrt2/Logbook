```python
import easyocr
from PIL import Image
import matplotlib.pyplot as plt
import cv2

# Load the OCR reader
reader = easyocr.Reader(['en', 'tl'])

# Path to the image
image_path = 'C:/Users/Bert/Documents/Thesis/Thesis Application-April 23, 2025 - No ocr but working buttons and table/IMG_1764.jpg'

# Read the text from the image
results = reader.readtext(image_path)

# Load the image using OpenCV (to allow drawing)
image = cv2.imread(image_path)

# Loop through results and draw bounding boxes and text
print("Extracted Text:")
for (bbox, text, prob) in results:
    # Print extracted text
    print(f"- {text}")

    # Unpack the bounding box points
    top_left = tuple(map(int, bbox[0]))
    bottom_right = tuple(map(int, bbox[2]))

    # Draw rectangle around detected text
    cv2.rectangle(image, top_left, bottom_right, (0, 255, 0), 2)

    # Put text above the rectangle
    cv2.putText(image, text, (top_left[0], top_left[1] - 10), 
                cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 0, 0), 2)

# Save the output image
output_path = 'annotated_image.jpg'
cv2.imwrite(output_path, image)

print(f"\nAnnotated image saved to {output_path}")
```

```python
import easyocr
from PIL import Image
import matplotlib.pyplot as plt

# Load the OCR reader (you can set language to 'en' or a list like ['en', 'ch_sim'])
reader = easyocr.Reader(['en', 'tl'])

# Read the text from an image
results = reader.readtext('C:/Users/Bert/Documents/Thesis/Thesis Application-April 23, 2025 - No ocr but working buttons and table/IMG_1761.jpg')

# Print the results
for (bbox, text, prob) in results:
    print(f"Detected text: '{text}' (Confidence: {prob:.2f})")
```

```python
from transformers import TrOCRProcessor, VisionEncoderDecoderModel
from PIL import Image
import torch

# Load the pretrained model and processor (for printed text)
processor = TrOCRProcessor.from_pretrained("microsoft/trocr-base-printed", use_fast=True)
model = VisionEncoderDecoderModel.from_pretrained("microsoft/trocr-base-printed")

# Load your image (replace with your own file path)
image = Image.open("test.jpg").convert("RGB")
image = image.resize((384, 384))
# Preprocess image and run inference
pixel_values = processor(images=image, return_tensors="pt").pixel_values
generated_ids = model.generate(pixel_values)

# Decode predicted tokens into string
generated_text = processor.batch_decode(generated_ids, skip_special_tokens=True)[0]

print("Recognized Text:", generated_text)
```
